# Web Scraping — books.toscrape.com
Amazon blocks scraping without login. Using practice site with same structure (product, rating, price, stock).

## Step 1 — Imports

```python

import requests 
import pandas as pd
from bs4 import BeautifulSoup
```

## Step 2 — Config

```python
URL = 'https://books.toscrape.com/catalogue/page-{}.html'
PAGES = 3
```

## Step 3 — Fetch HTML

```python
lst = []
for i in range(1,PAGES+1):
    r = requests.get(URL.format(i))
    lst.append(r.text)
```

## Step 4 — Parse HTML

```python
soup = BeautifulSoup(''.join(lst),'lxml')
books = soup.select('article.product_pod')
```

## Step 5 — Extract Fields

```python
rows = []
for b in books:
    title  = b.h3.a['title']
    price  = b.select_one('.price_color').text.strip()
    rating = b.select_one('p.star-rating')['class'][1]
    stock  = b.select_one('.availability').text.strip()
    rows.append({'Title': title, 'Price': price, 'Rating': rating, 'Stock': stock})
```

## Step 6 — Build DataFrame

```python
df = pd.DataFrame(rows)
df.head()
```

## Step 7 — Save CSV

```python
out = 'reviews.csv'
df.to_csv(out, index=False)
```
